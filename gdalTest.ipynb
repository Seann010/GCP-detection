```python
from osgeo import gdal

# Type the exact file path to your drone photo here
image_path = "G:/Fontys/OL4/OL4 AI/tiled_GCP_data/189_0_0.jpg"

dataset = gdal.Open(image_path)

if dataset is None:
    print("Error: Could not open the image file. Check your file path.")
else:
    
    domains = dataset.GetMetadataDomainList()
    
    for domain in domains:
        raw_meta = dataset.GetMetadata(domain)
        
        text_block = ""
        if isinstance(raw_meta, dict):
            text_block = "\n".join([f"{k}: {v}" for k, v in raw_meta.items()])
        elif isinstance(raw_meta, list):
            text_block = "\n".join(raw_meta)
        elif isinstance(raw_meta, str):
            text_block = raw_meta
            
        for line in text_block.split("\n"):
            line = line.strip()
            
            # 1. Drone Model
            if "ProductName" in line:
                print(line)
                
            # 2. GPS Location Coordinates
            elif "GpsLatitude" in line or "GpsLongitude" in line:
                print(line)
                
            # 3. Heights (takeoff and sea-level)
            elif "RelativeAltitude" in line or "AbsoluteAltitude" in line:
                print(line + " meters")
                
            # 4. Camera Box Angles (Gimbal)
            elif "Gimbal" in line and "Degree" in line:
                print(line + " degrees")
                
            # 5. Drone Body Angles (Flight)
            elif "Flight" in line and "Degree" in line:
                print(line + " degrees")
```
